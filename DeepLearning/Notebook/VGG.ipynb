```python
import torch
from torch import nn
from d2l import torch as d2l
```

```python
def vgg_block(num_conv, input_channel, out_channel):
    block = []
    for _ in range(num_conv):
        block.append(nn.Conv2d(input_channel, out_channel, kernel_size=3, padding=1))
        block.append(nn.ReLU())
        input_channel = out_channel
    block.append(nn.MaxPool2d(kernel_size=2, stride=2))
    return nn.Sequential(*block)
```

```python
def VGG(conv_array):
    blocks = []
    input_channel = 1
    for (num_conv, out_channel) in conv_array:
        blocks.append(vgg_block(num_conv, input_channel, out_channel))
        input_channel = out_channel
    return nn.Sequential(
        *blocks,
        nn.Flatten(),
        nn.Linear(out_channel * 7 * 7, 4096), 
        nn.ReLU(), 
        nn.Dropout(0.5),
        nn.Linear(4096, 4096), 
        nn.ReLU(), 
        nn.Dropout(0.5),
        nn.Linear(4096, 10)
        )
```

```python
conv_array = ((1, 64), (1, 128), (2, 256), (2, 512), (2, 512))
vgg_net = VGG(conv_array)
loss = nn.CrossEntropyLoss()
lr = 0.05
batch_size = 128
num_epoch = 10
train_iter, test_iter = d2l.load_data_fashion_mnist(batch_size) 
```
