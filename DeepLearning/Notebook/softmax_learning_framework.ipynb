**本代码通过调用 torch 已有的深度学习框架，简洁的实现softmax线性回归模型**

```python
#导入包
import torch
from torch import nn
from d2l import torch as d2l
```

```python
#样本准备
batch_size = 256
train_iter, test_iter = d2l.load_data_fashion_mnist(batch_size)
```

```python
#损失函数
#reduction = none 代表直接返回损失组成的张量而不进行求平均或求和的操作
loss = nn.CrossEntropyLoss(reduction='none')
```

```python
#初始化模型
#由于 torch 框架不会自动展平图片，所以这里要加一层由于展平图片的神经网络
net = nn.Sequential(nn.Flatten(), nn.Linear(784, 10))

def init_weights(m):
    #仅对线性层进行初始化
    if type(m) == nn.Linear:
        nn.init.normal_(m.weight, std=0.01)
#这里采用正态分布初始化权重
net.apply(init_weights)
```

```python
#训练方式
trainer = torch.optim.SGD(net.parameters(), lr=0.1)
#训练次数
num_epochs = 10
d2l.train_ch3(net, num_epochs, train_iter, test_iter, loss, trainer)
```
