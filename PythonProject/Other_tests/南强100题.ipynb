```python
from math import sqrt
def distance(x,y):
    line, d_1 = x + y, 0
    for i in range(0,line):
        d_1 += sqrt((0+i)**2 + (1+i)**2)
    d_2 = (x + (line * (line - 1)) / 2) * sqrt(2)
    return d_1 + d_2

for i in range(0,int(input())):
    x1,y1,x2,y2 = map(int,input().split())
    print("%.3f"%abs(distance(x1,y1) - distance(x2,y2)))
```
